```python
# /mlbio_scratch/wen2/scMAE/xenium_preprocessed_indomain

# read files in the directory "/mlbio_scratch/wen2/scMAE/xenium_preprocessed_indomain"
import os
import pandas as pd
import torch
import matplotlib.pyplot as plt
import numpy as np
# Get all files in the directory
pt_files = os.listdir("/mlbio_scratch/wen2/scMAE/xenium_preprocessed_indomain")

# read and plot the embeddings from the pt files
embeddings = []
for file in pt_files[:5]:
    embedding = torch.load(os.path.join("/mlbio_scratch/wen2/scMAE/xenium_preprocessed_indomain", file))
    print(embedding.device)
    embedding = embedding.cpu().numpy()
    # print(embedding.shape)
    embeddings.append(embedding)
    
# plt.close()
embeddings = np.array(embeddings)
# print(embeddings[0].device)
# print(embeddings[0])

max_values = np.max(embeddings, axis=0)
min_values = np.min(embeddings, axis=0)

print(max_values)
print(min_values)

```

```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2, perplexity=30, random_state=42)
reduced = tsne.fit_transform(embeddings)

plt.scatter(reduced[:, 0], reduced[:, 1])
plt.title('2D Scatter Plot with t-SNE')
plt.xlabel('Dim 1')
plt.ylabel('Dim 2')
plt.show()
```
